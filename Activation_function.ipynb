## Sigmoid Function or Logistic Function

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):

    try:
        equation = 1 / (1 + math.exp(-z))
        return equation

    except OverflowError:
        return 0.0
```

```python
print(" จำลอง Limit เมื่อ z ->(เข้าใกล้) ∞ (บวกมหาศาล) ")
positive_values = [0, 2, 10, 50, 100]
for z in positive_values:
    prob = sigmoid(z)
    print(f"z = {z:>3}  --> ความน่าจะเป็น: {prob:.10f}")

print()
print("-"*50)
print("\n จำลอง Limit เมื่อ z ->(เข้าใกล้) -∞ (ลบมหาศาล) ")
negative_values = [-2, -10, -50, -100, -800]
for z in negative_values:
    prob = sigmoid(z)
    print(f"z = {z:>4}  --> ความน่าจะเป็น: {prob:.10f}")
```

```python
plt.figure(figsize=(10,6))

# 1. ปรับสเกล: ใช้ np.linspace สร้างตัวเลข 100 จุดระหว่าง -10 ถึง 10 เพื่อให้กราฟสมูทและเห็นความโค้ง
z_values = np.linspace(-10, 10, 100)
sigmoid_values = [sigmoid(z) for z in z_values]

# พล็อตกราฟ Sigmoid
plt.plot(z_values, sigmoid_values, color='red', linewidth=2.5, label='Sigmoid Function')

# 2. เพิ่มเส้นกำกับจาก Limits ที่เราคำนวณไว้
plt.axhline(y=0.5, color='green', linestyle='--', alpha=0.7)
plt.axvline(x=0, color='blue', linestyle='--', alpha=0.7)

# 1. ลูกศรชี้การลู่เข้า 0 (ฝั่งซ้าย)
plt.annotate(
    'Limit come to y=0 since z is very negative', 
    xy=(-9, 0.01),              # พิกัดปลายลูกศรชี้ไป (x, y)
    xytext=(-6, 0.2),           # พิกัดที่วางข้อความ (x, y)
    arrowprops=dict(facecolor='green', shrink=0.05, width=2, headwidth=10),
    fontsize=12,
    color='green',
    fontweight='bold'
)

# 2. ลูกศรชี้การลู่เข้า 1 (ฝั่งขวา)
plt.annotate(
    'Limit come to y=1 since z is very large', 
    xy=(9, 0.99),               # พิกัดปลายลูกศรชี้ไป (x, y)
    xytext=(3, 0.75),           # พิกัดที่วางข้อความ (x, y)
    arrowprops=dict(facecolor='blue', shrink=0.05, width=2, headwidth=10), 
    fontsize=12,
    color='blue',
    fontweight='bold'
)

plt.xlabel('z')
plt.ylabel('Sigmoid (z)')
plt.title('Sigmoid Function and its Limits (Asymptotes)')
plt.legend(loc='center right')
plt.grid(alpha=0.3)
plt.show()
```

## สรุป 
- 1.) เมื่อ z เข้าใกล้ ∞ คือค่าเป็นบวกมากขึ้นเรื่อยๆ จนทำให้ผลลัพธ์เข้าใกล้ 1 ซึ่งในทาง Probabilty หมายความมีความคล้ายเกือบ 100% เราจะนิยามให้ค่าเป็น 1 
- 2.) เมื่อ z เข้าใกล้ -∞ คือค่ายิ่งติดลบมากขึ้นเรื่อยๆ ทำให้ผลลัพธ์เข้าใกล้ 0 ซึ่งในทาง Pribablity หมายความไม่มีความคล้ายกันเลย 0% เรานิยามให้ค่านั้นเป็น 0
##### นี้จึงเป็นสรุปว่าการนำเอา Limit เข้ามาช่วยจะช่วยตีความหายของค่าในฟังก์ชันของ Sigmoid ง่ายขึ้นและใช้พลังงานน้อยลงในการประมวลผล
